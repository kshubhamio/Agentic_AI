```python
import os
from dotenv import load_dotenv
from openai import OpenAI
from IPython.display import Markdown, display
```

```python
load_dotenv(override=True)
```

```python
google_api_key = os.getenv("GOOGLE_API_KEY")
google_base_url = os.getenv("GOOGLE_BASE_URL")
```

```python
if google_api_key:
    print("Google API key exists")
else:
    print("Google api key not found")
```

```python
gemini = OpenAI(base_url= google_base_url, api_key= google_api_key)
```

```python
request = "Please generate a challenging, descriptive question designed to assess Intelligence Quotient (IQ). Respond only with the question."
messages = [{"role": "user", "content": request}]
```

```python
response = gemini.chat.completions.create(
    model="gemini-3.5-flash",
    messages=messages
)

question = response.choices[0].message.content

print(question)
```

```python
competitors = []
answers = []
```

```python
messages = [{"role": "user", "content": question}]
```

```python
model_name = "gemini-3.5-flash"

response = gemini.chat.completions.create(model= model_name, messages= messages)
answer = response.choices[0].message.content

display(Markdown(answer))
competitors.append(model_name)
answers.append(answer)
```

```python
model_name = "gemini-3-flash-preview"

response = gemini.chat.completions.create(model= model_name, messages= messages)
answer = response.choices[0].message.content

display(Markdown(answer))
competitors.append(model_name)
answers.append(answer)
```

```python
model_name = "gemini-2.5-flash"

response = gemini.chat.completions.create(model= model_name, messages= messages)
answer = response.choices[0].message.content

display(Markdown(answer))
competitors.append(model_name)
answers.append(answer)
```

```python
model_name = "gemini-3.1-flash-lite"

response = gemini.chat.completions.create(model= model_name, messages= messages)
answer = response.choices[0].message.content

display(Markdown(answer))
competitors.append(model_name)
answers.append(answer)
```

```python
model_name = "gemini-2.5-flash-lite"
response = gemini.chat.completions.create(model= model_name, messages= messages)
answer = response.choices[0].message.content

display(Markdown(answer))
competitors.append(model_name)
answers.append(answer)
```

```python
for competitor, answer in zip(competitors,answers):
    print(f"Competitor: {competitor}\n\nAnswer:\n{answer}\n\n")
```

```python
from markdown_pdf import MarkdownPdf,Section

pdf = MarkdownPdf(toc_level=0)

markdown_text = "# LLM Response Analysis\n\n"
markdown_text += f"**<u>Question</u>:** {question}\n\n---\n\n"
for competitor, answer in zip(competitors,answers):
    markdown_text += f"## Competitor: {competitor}\n**<u>Answer</u>:** {answer}\n\n---\n\n"

pdf.add_section(Section(markdown_text))
pdf.save("llm_response_analysis (gemini).pdf")
```
