```python
import os
from dotenv import load_dotenv
from openai import OpenAI
from IPython.display import Markdown, display
```

```python
load_dotenv(override=True)
```

```python
google_api_key = os.getenv("GOOGLE_API_KEY")
huggingface_api_key = os.getenv("HUGGINGFACE_API_KEY")

google_base_url = os.getenv("GOOGLE_BASE_URL")
huggingface_base_url = os.getenv("HUGGINGFACE_BASE_URL")
```

```python
if google_api_key:
    print(f"Google API Key exists and begins {google_api_key[:2]}")
else:
    print("Google api key not found")
if huggingface_api_key:
    print(f"HuggingFace API Key exists and begins {huggingface_api_key[:2]}")
else:
    print("HuggingFace api key not found")
```

```python
gemini = OpenAI(base_url= google_base_url, api_key= google_api_key)
hf =OpenAI(base_url= huggingface_base_url, api_key= huggingface_api_key)
```

```python
request = "Please propose a hard, challenging question to assess IQ. Respond only with the question."
request += "Answer only with the question, no explanation."
messages = [{"role": "user", "content": request}]
```

```python
response = gemini.chat.completions.create(
    model="gemini-3.5-flash",
    messages=messages
)

question = response.choices[0].message.content

print(question)
```
